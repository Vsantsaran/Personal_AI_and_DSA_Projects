## climbStairs

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        a = 0
        b = 1
        c = 1
        s = 0
        print(a, b, c, s, end='\n\n')

        for i in range(n):
            s = b
            b += a
            a = s
            print(a, b, s)
        return b
```

```python
sol = Solution()
sol.climbStairs(6)
```

## Edit Distance

```python
class Solution:
    def minDistance(self, word1, word2):
        dp = [[0 for _ in range(len(word1) + 1)] for _ in range(len(word2) + 1)]

        for i in range(1, len(word2)+1):
            dp[i][0] = i

        for i in range(1, len(word1)+1):
            dp[0][i] = i

        for i in range(1, len(word2)+1):
            for j in range(1, len(word1)+1):
                if word1[j-1] == word2[i-1]:
                    dp[i][j] = dp[i-1][j-1]
                else:
                    dp[i][j] = 1 + min(dp[i][j-1], dp[i-1][j], dp[i-1][j-1])

        return dp[-1][-1]
```

```python
sol = Solution()
word1 = "santu"
word2 = "basant"

sol.minDistance(word1, word2)
```

```python
def compute_prefix_function(pattern):
    m = len(pattern)
    pi = [0] * m
    j = 0  # Length of previous longest prefix suffix

    for i in range(1, m):
        while j > 0 and pattern[i] != pattern[j]:
            j = pi[j - 1]  # Backtrack to previous match

        if pattern[i] == pattern[j]:
            j += 1
            pi[i] = j  # Store prefix length

    return pi

pattern = "ABABAC"
print(compute_prefix_function(pattern))  # Output: [0, 0, 1, 2, 3, 0]
```
