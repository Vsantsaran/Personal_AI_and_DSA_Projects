```python
a = "abcfmnplqade"
b = "lmsbqnghqde"
```

```python
def lcs(dp, a, b, i, j):
    if a[i] == b[j]:
        dp[i][j] = 1 + lcs(dp[i+1:][j+1:], a, b)
    else:
        dp[i][j] = max(lcs(dp, a, b, i+1, j), lcs(dp, a, b, i, j+1))
    return dp[i][j]
```

```python
dp = [[0 for _ in range(len(a))] for _ in range(len(b))]
i, j = 0, 0
lcs(dp, a, b, , i, j)
```
