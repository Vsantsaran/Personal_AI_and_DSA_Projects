## Only the length of LMIS

```python
def length_of_LIS(arr):
    n = len(arr)
    dp = [1] * n
    for i in range(n):
        for j in range(i):
            if arr[j] < arr[i]:
                dp[i] = max(dp[i], dp[j] + 1)
    return dp
```

```python
dp = length_of_LIS([-4, 3, 5, 2, 8, 6, 7, 9, 15, 10, -2, 4, 6])
max(dp)
```

## Actual LMIS using MCT

```python
class Node:
    def __init__(self, val, is_dummy=False):
        self.val = val
        self.is_dummy = is_dummy
        self.children = []
    def __getattr__(self, node):
        return getattr(self.children, node)
    def __len__(self):
        return len(self.children)
    def __str__(self):
        return f'{self.val} -> {[node.val for node in self.children]}'
```

```python
root = Node(0, True)
root.append(Node(5))
root.append(Node(6))
print(root)
```
