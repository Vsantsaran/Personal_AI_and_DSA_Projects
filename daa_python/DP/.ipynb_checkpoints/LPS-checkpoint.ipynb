```python
s = "turboventilator"
```

```python
def L(s, i, j):
    # print(i, j)
    # if i > j:
    #     return
    global count
    count += 1
    if i == j:
        return 1
    if s[i] == s[j]:
        if i+1 == j:
            return 2
        return 2 + L(s, i+1, j-1)
    return max(L(s, i+1, j), L(s, i, j-1))
```

```python
%%time

count = 0
lps = []
cost = L(s, 0, len(s)-1)
print(f"cost: {cost}  steps: {count}\nlps: {"".join(lps)}")
```

```python
def L2(s, dp, i, j, lps):
    if dp[i][j]: 
        return dp[i][j]
    # print(i, j)
    global count
    count += 1
    if i == j:
        lps.append(s[i])
        return 1
    if s[i] == s[j]:
        lps.append(s[i])
        lps.append(s[j])
        if i+1 == j:
            return 2
        dp[i][j] = 2 + L2(s, dp, i+1, j-1, lps)
        return dp[i][j]
    dp[i][j] = max(L2(s, dp, i+1, j, lps), L2(s, dp, i, j-1, lps))
    return dp[i][j]
```

```python
%%time

count = 0
lps = []
dp = [[0 for _ in range(len(s))] for _ in range(len(s))]
cost = L2(s, dp, 0, len(s)-1, lps)
print(f"cost: {cost}  steps: {count}\nlps: {"".join(lps)}")
```

```python
def longestPalindromeSubseq(s):
    n = len(s)
    dp = [[0] * n for _ in range(n)]

    # Base case: Single character substrings
    for i in range(n):
        dp[i][i] = 1

    # Fill the table diagonally
    for length in range(2, n + 1):  # Substring length
        for i in range(n - length + 1):
            j = i + length - 1
            if s[i] == s[j]:
                dp[i][j] = 2 + dp[i + 1][j - 1]
            else:
                dp[i][j] = max(dp[i + 1][j], dp[i][j - 1])

    # Step 2: Reconstruct LPS
    lps = [""] * dp[0][n-1]  # Placeholder for LPS
    i, j = 0, n - 1
    left, right = 0, dp[0][n-1] - 1  # Two pointers for building LPS

    while i <= j:
        if s[i] == s[j]:
            lps[left], lps[right] = s[i], s[j]  # Place at both ends
            left += 1
            right -= 1
            i += 1
            j -= 1
        elif dp[i + 1][j] > dp[i][j - 1]:  # Move to the larger value
            i += 1
        else:
            j -= 1

    return "".join(lps)  # Return the reconstructed LPS

# Example usage:
s = "turboventilator"
print(longestPalindromeSubseq(s))  # Output: "bbbb"
```
