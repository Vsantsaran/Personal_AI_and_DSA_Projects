```python
class Solution:
    def uniquePathsWithObstacles(self, G):
        dp = [[0 for _ in range(len(G[0]))] for _ in range(len(G))]

        dp[0][0] = -1 if G[0][0] else 1
        
        for i in range(len(G[0])):
            if G[0][i]:
                while i < len(G[0]):
                    dp[0][i] = -1
                    i += 1
                break
            else:
                dp[0][i] = 1

        for i in range(len(G)):
            if G[i][0]:
                while i < len(G) :
                    dp[i][0] = -1
                    i += 1
                break
            else:
                dp[i][0] = 1
        
        for i in range(1, len(G)):
            for j in range(1, len(G[0])):
                if G[i][j]:
                    dp[i][j] = -1
                elif dp[i][j-1] == -1 and dp[i-1][j] == -1:
                    dp[i][j] = -1
                    continue
                elif dp[i][j-1] == -1:
                    if dp[i-1][j] != -1:
                        dp[i][j] = dp[i-1][j]
                elif dp[i-1][j] == -1:
                    if dp[i][j-1] != -1:
                        dp[i][j] = dp[i][j-1]
                elif not G[i][j]:
                    dp[i][j] = dp[i-1][j] + dp[i][j-1]
        return 0 if dp[-1][-1] == -1 else dp[-1][-1], dp
```

```python
sol = Solution()
G = [[0, 0], [1, 1], [0, 0]]
sol.uniquePathsWithObstacles(G)
```

## Longest Palindromic Substring

```python
def LPSub(s, i, j, dp):
    if i == j:
        return 1
    if i+1 == j:
        if s[i] == s[j]:
            return 2
        return 0
    if i and j < len(s)-1:
        if s[i-1] == s[j+1]:
            if s[i] != s[j]:
                return 0
            dp[i][j] = LPSub(s, i+1, j-1, dp)
            return dp[i][j]
    if s[i] != s[j]:
        dp[i][j] = max(LPSub(s, i+1, j, dp), LPSub(s, i, j-1, dp))
        return dp[i][j]
    dp[i][j] = LPSub(s, i+1, j-1, dp)
    return max(dp)
```

```python
def findIndeces(arr, s_val):
    if s_val not in arr:
        return -1, -1
    a = arr.index(s_val)
    if a != len(arr)-1:
        if arr[a+1] == s_val:
            return a, a+1
    return a, a
```
