```python
a = "yckertgoosingys"
b = "boycutisverygoodnowadays"
```

```python
def lcs(dp, a, b, i, j):
    if i == len(a) or j == len(b):
        return 0
    
    global count
    count += 1
    
    if a[i] == b[j]:
        dp[i][j] = 1 + lcs(dp, a, b, i+1, j+1)
    else:
        dp[i][j] = max(lcs(dp, a, b, i+1, j), lcs(dp, a, b, i, j+1))
    return dp[i][j]
```

```python
%%time

count = 0
dp = [[0 for _ in range(len(b))] for _ in range(len(a))]
i, j = 0, 0
lcs(dp, a, b, i, j), count
```

```python
def lcsDP(dp, a, b, i, j):
    if i == len(a) or j == len(b):
        return 0
    if dp[i][j]:
        return dp[i][j]

    global count
    count += 1

    if a[i] == b[j]:
        dp[i][j] = 1 + lcsDP(dp, a, b, i+1, j+1)
    else:
        dp[i][j] = max(lcsDP(dp, a, b, i+1, j), lcsDP(dp, a, b, i, j+1))
    return dp[i][j]
```

```python
%%time
a = "yckertgoosingys"
b = "boycutisverygoodnowadays"

count = 0
dp = [[0 for _ in range(len(b))] for _ in range(len(a))]
i, j = 0, 0
lcsDP(dp, a, b, i, j), count
```
