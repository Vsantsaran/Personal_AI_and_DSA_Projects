```python
P = [(0, 0), (4, 5), (2, 1), (2, 2), (3, 6), (3, 2.5), (2, 4), (-4, 6), (4, 4), (1, 8), (5, 6), (0, 8)]
findConvexHull(P)
```

```python
def findConvexHull(P):
    x_min, y_min, x_max, y_max = findExtremes(P)
    extremes = (x_min, y_min, x_max, y_max)
    CHP = [*extremes]

    for point in P:
        if point in extremes:
            continue
        x, y = point
        X, Y = findAssociatedExtremePoints(extremes, point)
        # 'o' : for the point p to be in the convex hull - it should be outside the line
        # 'i' : for the point p to be in the convex hull - it should be inside the line
        sign = 'o' if Y == y_max else 'i'
        m = (Y[1] - X[1]) / (Y[0] - X[0])
        c = Y[1] - m*Y[0]
        line_func = lambda x: m*x + c
        gap = y - line_func(x)
        # CHP.append(point) if (sign == 'o' and gap > 0) or (sign == 'i' and gap < 0)
        
        if sign == 'o' and gap > 0:
            CHP.append(point)
        elif sign == 'i' and gap < 0:
            CHP.append(point)

    return CHP
```

```python
def findExtremes(points):
    sorted_on_x = sorted(points, key=lambda x: x[0])
    x_min, x_max = sorted_on_x[0], sorted_on_x[-1]

    sorted_on_y = sorted(points, key=lambda x: x[1])
    y_min, y_max = sorted_on_y[0], sorted_on_y[-1]

    return x_min, y_min, x_max, y_max
```

```python
def findAssociatedExtremePoints(extremes, point):
    x_min, y_min, x_max, y_max = extremes
    closest_points = []
    if (abs(x_min[0] - point[0]) > abs(x_max[0] - point[0])):
        closest_points.append(x_max)
    else:
        closest_points.append(x_min)

    if (abs(y_min[1] - point[1]) > abs(y_max[1] - point[1])):
        closest_points.append(y_max)
    else:
        closest_points.append(y_min)

    return closest_points
```
