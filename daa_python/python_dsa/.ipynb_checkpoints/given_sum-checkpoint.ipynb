```python
import numpy as np
```

```python
def two_sum_sorted(arr, target):
    left, right = 0, len(arr) - 1
    
    while left < right:
        current_sum = arr[left] + arr[right]
        if current_sum == target:
            return (left, right)
        elif current_sum < target:
            left += 1
        else:
            right -= 1
    
    return None

arr = [-23, -12, -3, -2, 3, 5, 8, 19, 23, 25, 31, 42, 63, 444]
s = -35
print(two_sum_sorted(arr, s))
```
