```python
def postfix_conversion(expr, stk):
    operators = ['*', '.', '%', '^', '+', '-', '/']
    brackets = ['(', ')', '{', '}', '[', ']']
    for i in expr:
        if i in brackets:
            continue
        if i in operators:
            sec = stk.get()
            fst = stk.get()
            val = eval(f'{fst}{i}{sec}')
            stk.put(val)
        else:
            conv = int(i)
            if conv in range(0, 10):
                stk.put(i)
    return stk.get()
```
