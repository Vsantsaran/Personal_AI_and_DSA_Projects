# Cycle Detection Using DFS and BFS in an Undirected Graph

```python
class Graph:
    def __init__(self, n):
        self.n = n
        self.q = queue.Queue(maxsize=n)
        self.stack = queue.LifoQueue(maxsize=n)
        # self.visited = [0 for _ in range(n)]
        self.flags = [-1 for _ in range(n)]
        
        # self.adj_mat = np.zeros((n, n), dtype=int)
        self.adj_mat = [[0 for _ in range(n)] for _ in range(n)]

    def add_edge(self, a, b):
        self.adj_mat[a][b] = self.adj_mat[b][a] = 1
```

```python
def detect_cycle_BFS(G, s=0):
    # BFS implementation - using a Queue 
    # check the neighbours of s, if any one of them is 0 in the flags array, a cycle is detected
    # if the value is 1 then don't enqueue
    # if the value is -1 then enqueue
    print(s, end=' ')
    # G.q.put(s)
    G.flags[s] = 1
    # G.visited[s] = 1
    for i, neighbour in enumerate(G.adj_mat[s]):
        if neighbour:
            if G.flags[i] == 0:
                print(i, end=' ')
                return True
            if G.flags[i] == -1:
                G.q.put(i)
                G.flags[i] = 0
    if G.q.empty():
        return False
    popped = G.q.get()
    return detect_cycle_BFS(G, popped)
```

```python
def detect_cycle_DFS(G, s=0):
    # DFS implementation - using a Stack 
    # check the neighbours of s, if any one of them is 0 in the flags array, a cycle is detected
    # if the value is 1 then don't enqueue
    # if the value is -1 then enqueue
    # print(G.flags, G.stack.queue, end='\n\n')
    print(s, end=' ')
    # G.q.put(s)
    G.flags[s] = 1
    # G.visited[s] = 1
    for i, neighbour in enumerate(G.adj_mat[s]):
        if neighbour:
            if G.flags[i] == 0:
                print(i, end=' ')
                return True
            if G.flags[i] == -1:
                G.stack.put(i)
                G.flags[i] = 0
    if G.stack.empty():
        return False
    popped = G.stack.get()
    return detect_cycle_DFS(G, popped)
```

```python
N = 5
G = Graph(N)
G.add_edge(0, 2)
G.add_edge(0, 3)
G.add_edge(1, 3)
G.add_edge(2, 4)
G.add_edge(3, 4)
if detect_cycle_BFS(G, 4):
    print('Cycle Detected')
```
