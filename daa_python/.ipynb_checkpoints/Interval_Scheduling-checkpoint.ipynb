```python
new = sorted(intervals, key=lambda x: (x[1], x[0]))
new
```

```python
def selctIntervals(arr):
    start = arr[0]
    for i in range(1, len(arr)):
        if arr[i][1] <= start[1] or arr[i][0] <= start[1]:
            continue
        
```
